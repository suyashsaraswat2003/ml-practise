```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
data = pd.read_csv("Concrete_Data.csv")

X = data[["Age (day)"]].values
y = data["Concrete compressive strength(MPa, megapascals) "].values

print("Dataset Shape:", data.shape)
print("Input Feature: Age (days)")
print("Target: Concrete Compressive Strength (MPa)")
print("First 5 Input Values:", X[:5].flatten())
print("First 5 Target Values:", y[:5])
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
print("Model Training Completed")
```

```python
model.fit(X_train, y_train)
print("Model Training Completed")
```

```python
y_pred = model.predict(X_test)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Actual Values:", y_test[:10])
print("Predicted Values:", y_pred[:10])
```

```python
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

```python
plt.scatter(X_test, y_test, label="Actual Values", alpha=0.6)

sorted_indices = np.argsort(X_test[:, 0])
plt.plot(
    X_test[sorted_indices, 0],
    y_pred[sorted_indices],
    color="red",
    label="Regression Line"
)

plt.xlabel("Concrete Age (days)")
plt.ylabel("Compressive Strength (MPa)")
plt.title("Simple Linear Regression Using Concrete Dataset")
plt.legend()
plt.show()
```
